```python
import json
import pandas as pd
```

```python

with open(file_path, 'r', encoding='utf-8') as file:
    data = json.load(file)

rows = []

for entry in data:
    professional_skills = entry.get("professional_skills") or []
    personal_skills = entry.get("personal_skills") or []
    
    professional_skills = professional_skills[:10] + [""] * (10 - len(professional_skills))
    personal_skills = personal_skills[:10] + [""] * (10 - len(personal_skills))
    
    row = {
        "url": entry["url"],
        "name": entry["name"],
        "level": entry["level"],
        "views": entry["views"],
        "company": entry["company"],
        "industry": entry["industry"],
        "geo": entry["geo"],
        "time": entry["time"],
        "Employment term": entry["Employment term"],
        "Category": entry["Category"],
        #"description": entry["description"],
        #"responsibilities": entry["responsibilities"],
        **{f"professional_skills_{i+1}": skill for i, skill in enumerate(professional_skills)},
        **{f"personal_skills_{i+1}": skill for i, skill in enumerate(personal_skills)},
    }
    rows.append(row)

df = pd.DataFrame(rows)

output_csv_path = 'parsing_results_19.csv'
df.to_csv(output_csv_path, index=False, encoding='utf-8')

print(f"Data successfully saved to file {output_csv_path}")
```
